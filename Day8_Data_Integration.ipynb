# Data Integration

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "ID": [1,2,3],
    "Name": ["Ali","Ahmed","Sara"]
})
df2 = pd.DataFrame({
    "ID": [1,2,3],
    "Marks":[80,90,85]
})
# AB humy in dono dataset ko combine karna hai 
merged_df = pd.merge(df1,df2,on="ID")
merged_df
```

```python
merged_df = pd.merge(df1,df2,on="ID",how= "inner")
merged_df
```

# Example 2 

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Ali", "Ahmed", "Sara", "Usman"]
})
df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 85]
})
merged_df = pd.merge(df1,df2,on="ID",how="inner")
merged_df
```

```python
merged_df = pd.merge(df1,df2,on="ID",how="left")
merged_df
```

```python
merged_df = pd.merge(df1,df2,on="ID",how="right")
merged_df
```

```python
merged_df = pd.merge(df1, df2, on="ID", how="outer")
merged_df
```

```python
# Concatination 
df1 = pd.DataFrame({
    "Name": ["Ali", "Ahmed"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Sara", "Usman"],
    "Marks": [85, 95]
})
combined_df = pd.concat([df1, df2], axis=0)
combined_df
```

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"]
})

df2 = pd.DataFrame({
    "Marks": [80, 90, 85]
})
combined_df = pd.concat([df1, df2], axis=1)
combined_df
```

```python
combined_df = pd.concat(
    [df1, df2],
    axis=0,
    ignore_index=True
)
combined_df
```
