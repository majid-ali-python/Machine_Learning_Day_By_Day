```python
import pandas as pd
import numpy as np
data = {
    "Marks" :[10,12,13,15,80,82,85,90,92]
}
df = pd.DataFrame(data)
df
# Data ko 3 bin ma divide karna 
df["Bin"] = pd.cut(df["Marks"],bins=3)
df
df["Smoothed_Median"] = df.groupby(
    "Bin", observed=True
)["Marks"].transform("median")

df
```

```python
def smoothing_boundaries(group):
    lower = group.min()
    upper = group.max()

    return group.apply(
        lambda x: lower if abs(x - lower) <= abs(x - upper) else upper
    )

df["Smoothed_Boundaries"] = df.groupby(
    "Bin", observed=True
)["Marks"].transform(smoothing_boundaries)

df
```
