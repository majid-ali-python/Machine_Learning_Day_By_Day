```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Department": ["IT", "HR", "IT", "Finance"]
}
df = pd.DataFrame(data)
df
```

```python
le = LabelEncoder()
df["Department"] = le.fit_transform(df["Department"])
df
```

```python
data = {
    "Student": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Grade": ["B", "A", "C", "B", "A"]
}
# sirf Gender par labelencoding karni hai 
df = pd.DataFrame(data)
df
le = LabelEncoder()
df["Grade"] = le.fit_transform(df["Grade"])
df
```

# Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Level": ["Low", "High", "Medium", "Low"]
}

df = pd.DataFrame(data)
df
oe = OrdinalEncoder(
    categories =[["Low","Medium","High"]]
)
df["Level"] = oe.fit_transform(df[["Level"]])
df
```
