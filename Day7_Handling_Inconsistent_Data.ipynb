# Handling Extra SPace


```python
data = {
    "Name": [" Ali ", "Ahmed  ", "  Sara", "Usman"],
    "City": [" Lahore ", "Karachi", " Lahore", "Islamabad"]
}

df = pd.DataFrame(data)
df
df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip()
df["City"] = df["City"].str.replace("Lahore", "LAHORE")
df
```
