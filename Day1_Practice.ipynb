```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Ahmed", "Usman", "Hassan", "Akram",
             "Bilal", "Zain", "Umar"],

    "Age": [20, 21, np.nan, 23, 22, np.nan, 24, 21],

    "Marks": [80, np.nan, 75, 90, 85, 70, np.nan, 80],

    "City": ["Lahore", "Karachi", "Lahore", np.nan,
             "Islamabad", "Karachi", np.nan, "Lahore"],

    "Fees": [5000, 6000, np.nan, 7000, 6500, 5000, np.nan, 6000]
}

df = pd.DataFrame(data)

# Q1: Complete dataset ko print karna
df
# Q2: isnull() sy missing values ko print karna
df.isnull()
# Q3: isnull().sum() sy har column ki missing value ko count karo
df.isnull().sum()
# Q4: info() say information ko check karna
df.info()
#5: Marks column ka mean calculate karna
mean_marks = df["Marks"].mean()
mean_marks
#6: Age k column ka median calculate karna
median_age = df["Age"].median()
median_age
#7: city column ka mode calculate karna
mode_city = df["City"].mode()
mode_city
#8: marks ki missing value mean say fill karna
df["Marks"] = df["Marks"].fillna(mean_marks)
df["Marks"]
#9: age ki missing value median say fill karo
df["Age"] = df["Age"].fillna(median_age)
df["Age"]
#10: city ki missing value mode sy fill karna
df["City"] = df["City"].fillna(mode_city)
df["City"]
#11: final dataset ko print karo
df
#12: dubara isnull().sum() laga kar check karo mising value nahi hai kia
df.isnull().sum()
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Ahmed", "Usman", "Hassan", "Akram",
             "Bilal", "Zain", "Umar"],

    "Age": [20, 21, np.nan, 23, 22, np.nan, 24, 21],

    "Marks": [80, np.nan, 75, 90, 85, 70, np.nan, 80],

    "City": ["Lahore", "Karachi", "Lahore", np.nan,
             "Islamabad", "Karachi", np.nan, "Lahore"],

    "Fees": [5000, 6000, np.nan, 7000, 6500, 5000, np.nan, 6000]
}

df = pd.DataFrame(data)

# Q1: Complete dataset ko print karna
df
# Q2: isnull() sy missing values ko print karna
df.isnull()
# Q3: isnull().sum() sy har column ki missing value ko count karo
df.isnull().sum()
# Q4: info() say information ko check karna
df.info()
#5: Marks column ka mean calculate karna
mean_marks = df["Marks"].mean()
mean_marks
#6: Age k column ka median calculate karna
median_age = df["Age"].median()
median_age
#7: city column ka mode calculate karna
mode_city = df["City"].mode()[0]
mode_city
#8: marks ki missing value mean say fill karna
df["Marks"] = df["Marks"].fillna(mean_marks)
df["Marks"]
#9: age ki missing value median say fill karo
df["Age"] = df["Age"].fillna(median_age)
df["Age"]
#10: city ki missing value mode sy fill karna
df["City"] = df["City"].fillna(mode_city)
df["City"]
#11: final dataset ko print karo
df
#13: fees ko b fill karo median say
df["Fees"] = df["Fees"].fillna(df["Fees"].median())
df["Fees"]
#12: dubara isnull().sum() laga kar check karo mising value nahi hai kia
df.isnull().sum()
```
