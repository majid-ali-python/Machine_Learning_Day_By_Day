```python
import pandas as pd

data = {
    "Salary": [20000, 22000, 25000, 27000, 30000,
               32000, 35000, 400000]
}

df = pd.DataFrame(data)

# Q1: Q1 calculate karo
Q1 = df["Salary"].quantile(0.25)
# Q2: Q3 calculate karo
Q3 = df["Salary"].quantile(0.75)
# Q3: IQR calculate karo
IQR = Q3 - Q1
# Q4: Lower Limit calculate karo
Lower_Limit = Q1 - 1.5*IQR
# Q5: Upper Limit calculate karo
Upper_Limit = Q3 + 1.5*IQR
# Q6: Outliers find karo
outlier = df[
    (df["Salary"]< Lower_Limit) |
    (df["Salary"]> Upper_Limit)
]
outlier
```

# Exercise Z_Score

```python
import pandas as pd

data = {
    "Salary": [20000, 22000, 25000, 27000,
               30000, 32000, 35000, 400000]
}

df = pd.DataFrame(data)

# Q1: Mean calculate karo
mean = df["Salary"].mean()
# Q2: Standard Deviation calculate karo
std = df["Salary"].std()
# Q3: Z-Score calculate karo
df["Z_Score"] = (df["Salary"]-mean)/std
# Q4: Z-Score > 3 ya Z-Score < -3
# se possible outliers find karo
outlier = df[
    (df["Z_Score"]>3) | (df["Z_Score"]<-3)
]
outlier
```

```python
print(df)
print(outlier)
```

# Outlier capping 

```python
# capping ka matlab hota hai k outlier ko delere karny ki bajay ya isy list ma adjust kar lata hai jasy:
data ={
    "Salary":[20000,22000,25000,27000,30000,40000]
}
# ab yaha par 40000 outlier hai to capping k zariya isy hum adjust karty hai 
# sab say pehly IQR nikaly gay 
df = pd.DataFrame(data)
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5*IQR
upper_limit = Q3 + 1.5*IQR
# Ab capping karty hai 
df["Salary"] = df["Salary"].clip(
    lower = lower_limit,
    upper = upper_limit
)
print(df)
```
