```python
data = {
    "Name": ["Ali", "Ahmad", "Ali", "Usman", "Ahmad"],
    "Age" : [20,22,20,21,22],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi"]
}
```

```python
import pandas as pd

data = {
    "Student": ["Ali", "Ahmed", "Usman", "Ali", "Bilal",
                "Ahmed", "Hassan", "Usman"],

    "Age": [20, 21, 22, 20, 23, 21, 24, 22],

    "Marks": [80, 75, 90, 80, 85, 75, 88, 90]
}

df = pd.DataFrame(data)

# Q1: Complete dataset print karo
df
# Q2: duplicated() se duplicate rows identify karo
df.duplicated()
# Q3: duplicated().sum() se total duplicates count karo
df.duplicated().sum()
# Q4: drop_duplicates() se duplicate rows remove karo
df=df.drop_duplicates()
# Q5: Final dataset print karo
df
# Q6: Dobara duplicates check karo
df.duplicated()
```
