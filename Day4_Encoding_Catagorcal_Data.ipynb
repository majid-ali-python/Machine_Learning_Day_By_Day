# One_Hot Encoding

```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
data = {
    "Name" : ["Ali", "Sara", "Ahmad", "Ayesha"],
    "Gender": ["Male", "Female", "Male", "Female"]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
encoded_df = pd.get_dummies(df,columns=["Gender"])
```

```python
import pandas as pd 
import numpy as np
```

```python
data = {
    "Name" : ["Ali", "Sara", "Ahmad", "Ayesha","Usman"],
    "City" : ["Lahore", "Karachi", "Lahore","Multan","Karachi"]
}
df = pd.DataFrame(data)
df
# Sirf city column par one hot encoding karna 
encoded_city = pd.get_dummies(df,columns=["City"],dtype=int)
encoded_city
```

# drop_first = Ture

```python
data = {
    "Name" : ["Ali", "Sara", "Ahmad", "Ayesha","Usman"],
    "City" : ["Lahore", "Karachi", "Lahore","Multan","Karachi"]
}
df = pd.DataFrame(data)
df
# Sirf city column par one hot encoding karna 
encoded_city = pd.get_dummies(df,columns=["City"],dtype=int, drop_first=True)
encoded_city
```
