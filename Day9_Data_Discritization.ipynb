```python
import pandas as pd

data = {
    "Age": [12, 16, 25, 40, 65]
}

df = pd.DataFrame(data)

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 12, 19, 59, 100],
    labels=["Child", "Teenager", "Adult", "Senior"]
)

df
```

# Equal width binning

```python
import pandas as pd

data = {
    "Marks": [10, 20, 30, 40, 50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)

df["Bins"] = pd.cut(df["Marks"], bins=3)

df
```

```python
df["Bins"] = pd.cut(df["Marks"], bins=5)

df
```

```python
df["Bins"] = pd.qcut(df["Marks"], q=3)

df
```
