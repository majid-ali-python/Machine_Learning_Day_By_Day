```python
import pandas as pd
import numpy as np

data2 = {
    "Student": ["Ali", "Ahmed", "Usman", "Hassan", "Akram"],
    "Marks": [80, 90, np.nan, 70, 60]
}

df = pd.DataFrame(data2)

# 1. Mean calculate karo
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df

# 2. Median calculate karo

# 3. Missing Marks ko Mean se fill karo

# 4. Final dataset print karo
```

```python
# 2 Median calculate karo
df["Marks"] = df["Marks"].median()
df["Marks"]
```

```python
# 3. Missing Marks ko Mean se fill karo
df2 = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(df2)
df["Marks"]
```
