# Normalization

```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import MinMaxScaler
data = {
    "Name" : ["Ali", "Ahmad", "Sara", "Ayesha", "Usman"],
    "Marks": [20,40,60,80,100]
}
df = pd.DataFrame(data)
df
scaler = MinMaxScaler()
df["Marks"] = scaler.fit_transform(df[["Marks"]])
df
```

```python
from sklearn.preprocessing import MinMaxScaler
data = {
    "Student": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Age": [15, 20, 25, 30, 35],
    "Marks": [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)
df
MMS = MinMaxScaler()
df["Marks"] = MMS.fit_transform(df[["Marks"]])
df
```

# Standardization 

```python
from sklearn.preprocessing import StandardScaler
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Marks": [10, 20, 30, 40, 50]
}

df = pd.DataFrame(data)
df
SS = StandardScaler()
df["Marks"] = SS.fit_transform(df[["Marks"]])
df
```
